```sql
CREATE OR REPLACE FUNCTION workspace.gold.get_dq_summary()
RETURNS TABLE (
    rule_id STRING,
    rule_name STRING,
    severity STRING,
    records_checked BIGINT,
    failures BIGINT,
    passes BIGINT,
    failure_pct DECIMAL(5,2)
)
LANGUAGE SQL
READS SQL DATA
RETURN
    SELECT
        rule_id,
        rule_name,
        severity,
        records_checked,
        failures,
        passes,
        CAST(failure_pct AS DECIMAL(5,2))
    FROM workspace.gold.dq_summary;
```
